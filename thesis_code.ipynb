```python
# note: python 3.11.9 used

path = "/Users/ael/Desktop/stats cheat sheet/"
# data prep
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# open data set
ic_scratch = pd.read_csv(path + "/Thesis Data Sets - MData.csv")
ic_scratch.dtypes
```

```python
ic_scratch['IC Ratio'] = ic_scratch['IC Ratio'].str.rstrip('%').astype(float) / 100
```

```python
# Plot
plt.figure(figsize=(15, 8))
sns.lineplot(
    x="Year", 
    y="IC Ratio", 
    hue="Country", 
    data=ic_scratch, 
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("IC Ratio")
plt.title("Income to Contribution Ratio Per Country (2015-2024)")
plt.legend(title="Region", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
